```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# we can apply the e^x function using numpy as follows
input_array = np.array([1,2,3]) # for multiple values (arrays)
exp_array = np.exp(input_array)
print(f"input to exp: {input_array}")
print(f"output to exp: {exp_array}")

input_val = 1
exp_val = np.exp(input_val)
print(f"input to exp: {input_val}")
print(f"output to exp: {exp_val}")
```

```python
def sigmoid(z):
    g = 1 / (1 + np.exp(-z))
    return g
```

```python
# let's use obove func
z_tmp = np.arange(-10,11) 
y = sigmoid(z_tmp)

# np.set_printoptions(precision=3)

print("Input (z),   Output (sigmoid(z))")
print(f"{np.c_[z_tmp, y]}")
```

```python
sigmoid(10000)
```

```python
plt.plot(z_tmp, y, c='b')
plt.title('Sigmoid function')
plt.xlabel('z')
plt.ylabel('sigmoid (z)')
plt.show()
```

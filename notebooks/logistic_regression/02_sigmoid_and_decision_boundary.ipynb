```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load our data
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]]) # 2 input features
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1)  # output

print(f"X: {X}, X.shape: {X.shape}")
print(f"y: {y}, y.shape: {y.shape}")
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# let's first plot our data and visualize it
# first we categorize our input rows based on two classes whether it's 0 or 1
negative = y.ravel() == 0 # it will select rows where y = 0
positive = y.ravel() == 1

plt.scatter(X[positive, 0], X[positive, 1], c='r', marker='x', lw=4, s=50, label='y=1')
plt.scatter(X[negative, 0], X[negative, 1], c='b', marker='o', lw=4, s=50, label='y=0')
plt.xlabel('$x_1$')
plt.ylabel('$y_1$')
plt.xlim(0,4)
plt.ylim(0,3.5)
plt.legend()
plt.show()
```

```python
z = np.arange(-10, 11)
g = sigmoid(z)

plt.plot(z, g, c='b')
plt.xlabel('z')
plt.ylabel('sigmoid (z)')
plt.show()
```

```python
# now plot the decision boundary of our assume parameters
negative = y.ravel() == 0 # it will select rows where y = 0
positive = y.ravel() == 1

x0 = np.arange(0,6)
x1 = 3 - x0

plt.scatter(X[positive, 0], X[positive, 1], c='r', marker='x', lw=4, s=50, label='y=1')
plt.scatter(X[negative, 0], X[negative, 1], c='b', marker='o', lw=4, s=50, label='y=0')
plt.plot(x0, x1, c='orange', label='decision boundary')

plt.xlabel('$x_1$')
plt.ylabel('$y_1$')
plt.xlim(0,4)
plt.ylim(0,3.5)
plt.legend()
plt.show()
```

```python
import math, copy
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=2)
```

```python
X_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])

print(f"X_train: {X_train}, X_train shape: {X_train.shape}")
print(f"y_train: {y_train}, y_train shape: {y_train.shape}")
```

```python
def predict_output_loop(x, w, b):
    n = x.shape[0]
    f_wb = 0
    for i in range(n):
        f_wb = f_wb + w[i] * x[i]
    f_wb += b
    return f_wb
```

```python
def predict_output_dot(x, w, b):
    f_wb = np.dot(x,w) + b
    return f_wb
```

```python
def compute_cost(X, y, w, b):
    m,n = X.shape
    cost_sum = 0
    for i in range(m):
        f_wb = np.dot(X[i], w) + b
        cost = (f_wb - y[i]) ** 2
        cost_sum += cost
    total_cost = cost_sum / (2 * m)
    return total_cost
```

```python
def compute_gradient(X, y, w, b):
    m,n = X.shape
    dj_dw = np.zeros(n)
    dj_db = 0
    for i in range(m):
        err = (np.dot(X[i],w) + b) - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] +  err * X[i][j]
        dj_db = dj_db + err
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    return dj_dw, dj_db
```

```python
def gradient_descent(X, y, w_in, b_in, alpha, num_iters, compute_cost, compute_gradient):

    J_history = []
    w = copy.deepcopy(w_in)
    b = b_in
    
    for i in range(num_iters):

        dj_dw, dj_db = compute_gradient(X, y, w, b)
    
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
    
        J_history.append(compute_cost(X, y, w, b))

        if i % math.ceil(num_iters/10) == 0:
            print(
                f"Iterations {i:4}: Cost: {J_history[-1]:.3e}",
                f"w: {w}, b: {b}"
            )

    return w, b, J_history
```

```python
# let's run it
w_init = np.zeros(X_train.shape[1])
b_init = 0
alpha = 1.0e-10
iterations = 10000

w_final, b_final, hist = gradient_descent(X_train, y_train, w_init, b_init, alpha, iterations, compute_cost, compute_gradient)

print(f"\n\nw, b found by gradient descent: {w_final}, {b_final}.")
```

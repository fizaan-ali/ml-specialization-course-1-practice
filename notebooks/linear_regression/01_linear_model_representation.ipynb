```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# x_train is the input variable or feature (size _ in 1000 square feet)
# y_train is the output variable or target (price _ in 1000s of dollars)
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print("x_train:", x_train)
print("y_train:", y_train)
```

```python
x_train.shape
```

```python
# m is the no. of training examples
m = x_train.shape[0]
print("No. of training examples:", m)
```

```python
# training example -> (x_i, y_i)
# x_i -> input feature, y_i -> corresponding target
i = 0
x_i = x_train[i]
y_i = y_train[i]

print(f"(x_{i}, y{i}): ({x_i}, {y_i})")
```

```python
# plot these points
plt.scatter(x_train, y_train, marker='x', color='r')
plt.title("Housing Prices")
plt.xlabel("Size (1000 square feet)")
plt.ylabel("Price (in 1000s of dollars)")
plt.show()
```

```python
w = 100
b = 100
print("w:", w)
print("b:", b)
```

```python
def compute_model_output(x, w, b):
    m = x.shape[0] # training set size
    f_wb = np.zeros(m) # empty array of m size
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
# now let's plot our model prediction
tmp_f_wb = compute_model_output(x_train, w, b)

plt.plot(x_train, tmp_f_wb, color='blue', label='our prediction')
plt.scatter(x_train, y_train, color='red', marker='x', label='actual values')
plt.title("Housing Prices")
plt.xlabel("Size (1000 square feet")
plt.ylabel("Price (in 1000s of dollars")
plt.legend()
plt.show()
```

```python
# as we can see above the line does not fit best for our training set. we have to put suitable values
# for w and b to get the best line
```

```python
import numpy as np
```

```python
X_train = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])
```

```python
X_train
```

```python
y_train
```

```python
positive = y_train == 1
negative = y_train == 0

print(f"positive: {positive}\nnegative: {negative}")
```

```python
X_train[positive] # select all rows with having y == 1
```

```python
X_train[negative] # all rows having y == 0
```

```python
X_train[positive, 0] # value in the 1st column of positive examples (having y == 1)
```

```python
X_train[negative, 1] # values in the 2nd column of negative examples (having y == 0)
```

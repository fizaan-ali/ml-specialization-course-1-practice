```python
def my_dot(a, b):
    x = 0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# now let's compare time between my_dot and vectorized dot provided by numpy
```

```python
a = np.random.rand(1000000) # very large arrays
b = np.random.rand(1000000)

tic = time.time() # returns the no. of seconds since epoch to current
c = np.dot(a,b)
toc = time.time()

print(f"np.dot(a,b) = {c:.4f}")
print(f"vectorized version duration = {1000*(toc-tic):.4f} ms")

tic = time.time()
c = my_dot(a,b)
toc = time.time()

print(f"my_dot(a,b) = {c:.4f}")
print(f"loop version duration = {1000*(toc-tic):.4f} ms")
```
